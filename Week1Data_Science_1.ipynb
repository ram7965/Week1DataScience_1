```python

# Customer Churn Prediction - Simple Example

# Collect customer information
name = input("What is the customer's name? ")
age = int(input("How old is the customer? "))
purchase_history = int(input("How many purchases have they made? "))
number_of_orders = int(input("How many orders have they placed? "))
spending_amount = float(input("How much have they spent? "))
days_since_purchase = int(input("How many days since their last purchase? "))
customer_service = int(input("How many customer-service interactions? "))

# Display customer information
print("\n--- Customer Information ---")
print("Name:", name)
print("Age:", age)
print("Purchase History:", purchase_history)
print("Number of Orders:", number_of_orders)
print("Spending Amount: $", spending_amount)
print("Days Since Last Purchase:", days_since_purchase)
print("Customer-Service Interactions:", customer_service)

# Determine churn risk
if days_since_purchase > 90 or customer_service >= 5:
    churn_risk = "HIGH"
elif days_since_purchase > 30:
    churn_risk = "MEDIUM"
else:
    churn_risk = "LOW"

print("\nCustomer Churn Risk:", churn_risk)
```
